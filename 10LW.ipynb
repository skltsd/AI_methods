```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Ice Cream.csv')
dataset.head()
```

```python
X = dataset.iloc[:, 0:1].values
y = dataset.iloc[:, 1].values
print("Матрица признаков (температура):")
print(X[:5])
print("Зависимая переменная (выручка):")
print(y[:5])
```

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, y)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=10)
X_poly = poly_reg.fit_transform(X)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X_poly, y)
```

```python
y_pred_lin = lin_reg.predict([[25]])
y_pred_poly = lin_reg_2.predict(poly_reg.fit_transform([[25]]))
print("Предсказание линейной модели:", y_pred_lin)
print("Предсказание полиномиальной модели:", y_pred_poly)
```

```python
plt.scatter(X, y, color='red')
plt.plot(X, lin_reg.predict(X), color='blue')
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('Temperature')
plt.ylabel('Revenue')
plt.show()
```

```python
plt.scatter(X, y, color='red')
plt.plot(X, lin_reg_2.predict(poly_reg.fit_transform(X)), color='blue')
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('Temperature')
plt.ylabel('Revenue')
plt.show()
```

```python
X_grid = np.arange(min(X), max(X), 0.1)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color='red')
plt.plot(X_grid, lin_reg_2.predict(poly_reg.fit_transform(X_grid)), color='blue')
plt.title('Truth or Bluff (Polynomial Regression) - smooth curve')
plt.xlabel('Temperature')
plt.ylabel('Revenue')
plt.show()
```
